```python
import pandas as pd

# READING COMPRESSION 2 -> 1
df = pd.read_csv("zahradnik_cont_speech_features.csv")

# Convert HC621a / HC621b -> HC621
df["sample"] = df["sample"].str.replace(r"[a-zA-Z]$", "", regex=True)

# Convert feature columns to numeric where possible
feature_cols = df.columns.drop(["sample", "group"])
df[feature_cols] = df[feature_cols].apply(pd.to_numeric, errors="coerce")

# Average duplicate samples
compressed = (
    df.groupby(["sample", "group"], as_index=False)[feature_cols]
      .mean()
)

compressed.to_csv("compressed_zahradnik_features.csv", index=False)
```

```python
import pandas as pd
# SUSTAINED VOWELS COMPRESSION 4 -> 1
df = pd.read_csv("phonation_features.csv")

df["sample"] = df["sample"].str.replace(r"(a1|a2|i1|i2|a|i)$", "", regex=True)

# Convert feature columns to numeric where possible
feature_cols = df.columns.drop(["sample", "group"])
df[feature_cols] = df[feature_cols].apply(pd.to_numeric, errors="coerce")

# Average duplicate samples
compressed = (
    df.groupby(["sample", "group"], as_index=False)[feature_cols]
      .mean()
)

compressed.to_csv("compressed_phonation_features.csv", index=False)
```
